```python
import requests
import pandas as pd
import os

# --- Configuration ---
# We are targeting the "Measles - number of reported cases" indicator
INDICATOR_CODE = "WHS3_41" 
API_ENDPOINT = f"https://ghoapi.azureedge.net/api/{INDICATOR_CODE}"

# --- Create data/raw directory if it doesn't exist ---
output_dir = '../data/raw'
os.makedirs(output_dir, exist_ok=True)
output_file = os.path.join(output_dir, 'who_measles_cases.csv')


# --- Fetch Data from WHO API ---
print(f"Fetching data for indicator {INDICATOR_CODE}...")
try:
    response = requests.get(API_ENDPOINT)
    response.raise_for_status()  # This will raise an error for bad responses (4xx or 5xx)
    
    # The API returns data in a JSON format
    data = response.json()
    
    # The actual data is nested under the 'value' key
    records = data['value']
    
    print(f"Successfully fetched {len(records)} records.")

    # --- Convert to Pandas DataFrame ---
    df = pd.DataFrame(records)
    
    print("Data converted to DataFrame. Here's a preview:")
    print(df.head())

    # --- Save the Raw Data ---
    df.to_csv(output_file, index=False)
    print(f"\nRaw data successfully saved to: {output_file}")

except requests.exceptions.RequestException as e:
    print(f"Error fetching data: {e}")
except KeyError:
    print("Error: 'value' key not found in the JSON response. The API structure may have changed.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
